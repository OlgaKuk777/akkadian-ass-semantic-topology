# **Аккадский язык**

*Кукушкина Ольга*

    Древнейший этап развития А.я. (примерно с середины III и до начала II тыс. до н. э.) называют староаккадским. 
    Начиная примерно с 1950 г. до н. э. вводится раздельная периодизация для ассирийского и вавилонского диалектов, в каждом из которых выделяют три периода — старый, средний и новый. Соответственно, по хронологическим и географическим критериям выделяются шесть основных форм А.я.:
    
     старовавилонский (1950–1530 гг. до н. э.) и староассирийский (1950–1750 гг. до н. э.),
     средневавилонский (1530–1000 гг. до н. э.) и среднеассирийский (1500–1000 гг. до н. э.),
     нововавилонский (1000–625 гг. до н. э.) и новоассирийский (1000–600 гг. до н. э.).
    Поскольку вавилонский диалект просуществовал дольше ассирийского, для него выделяется еще один, поздневавилонский, период (625 г. до н. э. — 100 гг. н. э.).

```python
import os
import json
import sys

def extract_lemmas_from_cdl(cdl_list, lemmas):
    """Рекурсивно извлекает 'cf' из структуры CDL"""
    for item in cdl_list:
        if item.get("node") == "l":  # лексема
            f_data = item.get("f", {})
            cf = f_data.get("cf")
            # Пропускаем неясные, повреждённые или служебные токены
            if cf and cf not in ("[unclear]", "[missing]", "u", "x"):
                lemmas.append(cf)
        elif "cdl" in item:
            extract_lemmas_from_cdl(item["cdl"], lemmas)

def process_corpusjson_dir(corpusjson_dir):
    """Обрабатывает все JSON-файлы в corpusjson/ и возвращает список предложений (списков лемм)."""
    sentences = []
    for filename in os.listdir(corpusjson_dir):
        if not (filename.startswith("P") and filename.endswith(".json")):
            continue

        filepath = os.path.join(corpusjson_dir, filename)
        try:
            with open(filepath, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception as e:
            print(f"Ошибка при чтении {filename}: {e}", file=sys.stderr)
            continue

        # Извлекаем все леммы из текста
        all_lemmas = []
        if "cdl" in data:
            extract_lemmas_from_cdl(data["cdl"], all_lemmas)
        
        if all_lemmas:
            # Весь текст как одно "предложение" (SBOW обычно так и делает)
            sentences.append(all_lemmas)
        else:
            print(f"Пустой текст: {filename}", file=sys.stderr)
    
    return sentences

if __name__ == "__main__":
    corpusjson_path = "/home/kuklelik/Jupyter/akk/saao/saa01/corpusjson"
    
    if not os.path.isdir(corpusjson_path):
        print(f"Папка не найдена: {corpusjson_path}", file=sys.stderr)
        sys.exit(1)

    print("Извлечение лемм...")
    sentences = process_corpusjson_dir(corpusjson_path)
    
    # Сохранение
    # одна строка = одно предложение, леммы через пробел
    output_file = "neo_assyrian_lemmatized1.txt"
    with open(output_file, "w", encoding="utf-8") as f:
        for lemmas in sentences:
            f.write(" ".join(lemmas) + "\n")
    
    print(f"ВсЁ! Сохранено {len(sentences)} документов в '{output_file}'")
```

```python
import os
import json
import sys
from collections import defaultdict


def extract_lemmas_and_lang_from_cdl(cdl_list, lemmas_by_lang):
    """Рекурсивно извлекает леммы и их lang, группируя по языку"""
    for item in cdl_list:
        if item.get("node") == "l":  # проверяем узел что это лексема l
            f_data = item.get("f", {}) # подсловарь features
            cf = f_data.get("cf")
            lang = f_data.get("lang", "unknown") #------------ язык!
            
            #Пропускаем мусор
            if not cf or cf in ("[unclear]", "[missing]", "u", "x"):
                continue
                
            # Сохраняем только аккадские варианты
            if lang and lang.startswith("akk-x-"):
                lemmas_by_lang[lang].append(cf)   # lang: cf
                
        elif "cdl" in item:
            extract_lemmas_and_lang_from_cdl(item["cdl"], lemmas_by_lang)  # рекурсия для вложенных штук

#             {
#   "akk-x-neoass": ["ardu", "ilu", "šarru", "ana", ...],
#   "akk-x-neobab": ["māru", "bēl", ...]
# }


def process_corpusjson_dir(corpusjson_dir):
    """Обрабатывает JSON-файлы и возвращает словарь: lang -> список документов (списков лемм)"""
    lang_documents = defaultdict(list)  # lang -> [ [лемма1, лемма2, ...], ... ] (тут для люб ключа пустой лист)

    if not os.path.isdir(corpusjson_dir): # если папка не существует
        return lang_documents

    for filename in os.listdir(corpusjson_dir): # перебор в папке
        if not (filename.startswith("P") and filename.endswith(".json")): # тип P123456.json
            continue

        filepath = os.path.join(corpusjson_dir, filename) # создаём путь к файлу P123456.json
        try:
            with open(filepath, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception as e:
            print(f"Ошибка при чтении {filepath}: {e}", file=sys.stderr)
            continue

        lemmas_by_lang = defaultdict(list) # временный пока что пустой словарь ДЛЯ КАЖДОГО файла
        if "cdl" in data:
            extract_lemmas_and_lang_from_cdl(data["cdl"], lemmas_by_lang)

        # Каждый документ добавляем как отдельную строку в соответствующий lang
        for lang, lemmas in lemmas_by_lang.items():
            if lemmas:
                lang_documents[lang].append(lemmas)  # весь список лемм для каждого языка в lang_documents
    
    return lang_documents # один JSON-файл = один документ = одна строка в выходном 

# {
#   "akk-x-neoass": [
#     ["ardu", "ilu", "šarru"],        #  P123456.json
#     ["ana", "bēlīya", "qibī"],       # P123457.json
#     ...
#   ],
#   "akk-x-neobab": [
#     ["māru", "bēl", "ṣalam"],        # P123477.json
#     ...
#   ]
# }

def collect_all_lang_documents(base_paths):
    """Собирает документы по языкам из списка путей"""  # то же что process_corpusjson_dir но по ВСЕМ ПАПКАМ ВМЕСТЕ
    total = defaultdict(list) #будет содержать все документы из всех папок
    for path in base_paths:
        docs_by_lang = process_corpusjson_dir(path) #документы из этой папки
        for lang, docs in docs_by_lang.items():
            total[lang].extend(docs)
    return total



if __name__ == "__main__":
    # Все пути для обработки
    all_paths = [f"/home/kuklelik/Jupyter/akk/saao/saa{str(i).zfill(2)}/corpusjson" # zfill 1 в "01", 10 в "10"
        for i in range(1, 22)] # + [..другие любые пути...]
#     all_paths = ["/home/kuklelik/Jupyter/akk/saao/saa01/corpusjson",
#                 "/home/kuklelik/Jupyter/akk/saao/saa04/corpusjson"]
    
    print("Извлечение лемм с учётом lang...")
    lang_docs = collect_all_lang_documents(all_paths)

    # какие lang встретились
    print("Найденные языковые метки:")
    for lang in sorted(lang_docs.keys()):
        print(f"  {lang}: {len(lang_docs[lang])} документов")

    mapping = {
        "akk-x-neoass": "neo_assyrian_lemmatized.txt",
        "akk-x-neobab": "neo_babylonian_lemmatized.txt",
        "akk-x-stdbab": "std_babylonian_lemmatized.txt",
        "akk-x-midass": "mid_assyrian_lemmatized.txt"
    }

    for lang_tag, filename in mapping.items():
        docs = lang_docs.get(lang_tag, []) # берёт доки для lang_tag (определённого языка)
        
        if docs: # если есть - записывает
            with open(filename, "w", encoding="utf-8") as f:
                for lemmas in docs:
                    f.write(" ".join(lemmas) + "\n")
            print(f"\n Сохранено {len(docs)} документов в '{filename}'")
        else:
            print(f"\n Нет данных для {lang_tag} - '{filename}' не создан")

    print("\nВСЁ!")
```

```python
from datetime import datetime
import os
import sys
import time
import traceback
from collections import Counter
from pathlib import Path
from urllib.parse import urlparse
import json
from playwright.async_api import async_playwright  #< АСИНХРОННЫЙ
from bs4 import BeautifulSoup
import asyncio


def extract_lemmatized_text(html: str) -> dict:
    soup = BeautifulSoup(html, 'html.parser')
    table = soup.find('table', class_='transliteration')
    if not table:
        raise ValueError("Таблица транслитерации не найдена")

    parts = []
    langs = []

    for td in table.select('td.tlit'):
        td_copy = BeautifulSoup(str(td), 'html.parser').find('td')
        if not td_copy:
            continue

        for span in td_copy.find_all('span', class_='w'):
            title = span.get('title', '')
            lemma = title.split('[', 1)[0].strip() if '[' in title else span.get_text().strip()
            if lemma:
                span.replace_with(lemma)

            for cls in span.get('class', []):
                if cls.startswith(('akk-x-')):
                    langs.append(cls)
                    break

        cell_text = td_copy.get_text().strip()
        if cell_text:
            parts.append(cell_text)

    text = ' '.join(parts)
    text = ' '.join(text.split())
    lang = Counter(langs).most_common(1)[0][0] if langs else None
    return {"text": text, "lang": lang}


def save_exception_log(savepath: str, err: Exception, source: str, **kwargs):
    print(f'\n-------  ◀ save_exception_log ▶  source={source}  savepath={savepath})  -------\n')
    os.makedirs(os.path.join(savepath, 'errors'), exist_ok=True)
    now = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
    filename = f"{source}_{now}.json"
    full_savepath = os.path.join(savepath, 'errors', filename)

    error_type = type(err).__name__
    error_message = str(err)
    _, _, tb = sys.exc_info()
    tb_line = traceback.extract_tb(tb)[-1]
    log_data = {
        "source": source,
        "type": error_type,
        "message": error_message,
        'filename': tb_line.filename,
        'lineno': tb_line.lineno,
        'code': tb_line.line,
        'traceback': traceback.format_exc(),
        "timestamp": datetime.now().isoformat(),
    }
    if kwargs:
        for name, val in kwargs.items():
            log_data[name] = val

    os.makedirs(os.path.dirname(full_savepath), exist_ok=True)
    with open(full_savepath, "w", encoding="utf-8") as f:
        json.dump(log_data, f, ensure_ascii=False, indent=2)
    print(f"Ошибка сохранена в {full_savepath}")


async def run(source_url, savepath_output, savepath_data):
    source_url = source_url.strip()
    corpus_name = urlparse(source_url).path.strip('/').split('/')[-2] + '.json'

    try:
        async with async_playwright() as p:
            browser = await p.chromium.launch(headless=True)
            page = await browser.new_page()
            await page.goto(source_url)
            await page.wait_for_selector("table", timeout=10000)

            # Получаем все строки таблицы
            rows = await page.query_selector_all("table tr")
            corpus = []

            for i in range(len(rows)):
                cells = await rows[i].query_selector_all("td")
                if len(cells) >= 2:
                    link = await cells[1].query_selector("a")
                    if link:
                        await link.click()
                        await page.wait_for_load_state("networkidle")

                        try:
                            html = await page.content()
                            data = extract_lemmatized_text(html)
                            corpus.append(data)
                            print('кайф')
                        except Exception as err:
                            print(f'Почему то получилась хуйня: {page.url}')
                            print(err)

                        # Возвращаемся на список
                        await page.goto(source_url)
                        await page.wait_for_selector("table", timeout=10000)
                        # Обновляем список строк (DOM мог измениться)
                        rows = await page.query_selector_all("table tr")

            savepath_corpus = os.path.join(savepath_output, corpus_name)
            print(f'Корпус сохранён в {savepath_corpus}')
            with open(savepath_corpus, 'w', encoding='utf-8') as file:
                json.dump(corpus, file, ensure_ascii=False, indent=4)

            await browser.close()

    except Exception as err:
        save_exception_log(savepath_data, err, corpus_name, **{'Оляша': "Милашка"})
        print(f"Ошибка при подключении или работе: {err}")


async def main():
    BASE_DIR = Path.cwd()
    savepath_data = BASE_DIR / 'data_html_akk'
    savepath_output = savepath_data / 'output'
    os.makedirs(savepath_output, exist_ok=True)
    
    targets = [
        "https://oracc.museum.upenn.edu/aemw/amarna/pager",
        "https://oracc.museum.upenn.edu/akklove/corpus",
        "https://oracc.museum.upenn.edu/glass/corpus",
    ]

    for url in targets:
        await run(url, savepath_output, savepath_data)
```

```python
import os
import json
import re
from collections import defaultdict
from pathlib import Path

# f очистки текста 
def clean_akkadian_text(text: str) -> list:
    """
    Удаляет мусор, но сохраняет аккадские буквы с диакритикой
    """
    if not text or not isinstance(text, str):
        return []
    
    # Удаляем всё внутри квадратных скобок
    text = re.sub(r'\[.*?\]', ' ', text)
    
    # Удаляем угловые скобки
    text = re.sub(r'\<\<.*?\>\>', ' ', text)
    
    # Удаляем специальные клинописные знаки
    text = re.sub(r'[⸢⸣]', ' ', text)
    
    # Удаляем цифры (включая подстрочные/надстрочные типа qa₂-be-ki)
    text = re.sub(r'\d+', ' ', text)
    
    # Разбиваем на токены
    tokens = text.split()
    clean_tokens = []
    
    for token in tokens:
        # Удаляем x-x-x и подобное
        if re.search(r'x-|-x', token.lower()):
            continue
        # Удаляем недопустимые символы с краёв
        token = re.sub(r"^[^a-zA-Z\u0100-\u024F\u1E00-\u1EFF\u0250-\u02AF]+", "", token)
        token = re.sub(r"[^a-zA-Z\u0100-\u024F\u1E00-\u1EFF\u0250-\u02AF]+$", "", token)
        
        # Пропускаем короткие и мусорные токены
        if len(token) < 2:
            continue
        
        # Оставляем только буквы + диакритику + апостроф/дефис внутри
        cleaned = re.sub(r"[^a-zA-Z\u0100-\u024F\u1E00-\u1EFF\u0250-\u02AF'-]", "", token)
        cleaned = cleaned.strip("'-")
        
        if len(cleaned) >= 2 and cleaned not in {"u", "x"}:
            clean_tokens.append(cleaned)
    
    return clean_tokens


# Обработка файлов из output
def main():
    BASE_DIR = Path("/home/kuklelik/Jupyter")
    INPUT_DIR = BASE_DIR / "data_html_akk" / "output"
    OUTPUT_DIR = BASE_DIR / "data_html_akk" / "output_final"
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    # Собираем документы по языкам
    lang_documents = defaultdict(list)

    for filename in INPUT_DIR.glob("*.json"):
        print(f"Обработка {filename}...")
        try:
            with open(filename, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception as e:
            print(f" Ошибка чтения: {e}")
            continue

        for item in data:
            if not isinstance(item, dict):
                continue
            text = item.get("text", "")
            lang = item.get("lang", "")
            
            if not lang or not lang.startswith("akk-x-") or not text:
                continue
                
            lemmas = clean_akkadian_text(text)
            if lemmas:
                lang_documents[lang].append(lemmas)

    mapping = {
        "akk-x-mbperi": "mid_babylonian_peripheral_lemmatized.txt",
        "akk-x-midass": "mid_assyrian_lemmatized.txt",
        "akk-x-stdbab": "std_babylonian_lemmatized.txt",
        "akk-x-oldbab": "old_babylonian_lemmatized.txt",
        "akk-x-oldakk": "old_akkadian_lemmatized.txt",
        "akk-x-neoass": "neo_assyrian_lemmatized.txt",
        "akk-x-neobab": "neo_babylonian_lemmatized.txt",
        "akk-x-midbab": "mid_babylonian_lemmatized.txt"
    }

    total_docs = 0
    for lang_tag, filename in mapping.items():
        docs = lang_documents.get(lang_tag, [])
        if docs:
            filepath = OUTPUT_DIR / filename
            with open(filepath, "w", encoding="utf-8") as f:
                for lemmas in docs:
                    f.write(" ".join(lemmas) + "\n")
            print(f" Сохранено {len(docs)} документов в '{filename}'")
            total_docs += len(docs)
        else:
            print(f" Нет данных для {lang_tag}")

    print(f" Всего обработано {total_docs} документов.")
    print(f" Результаты в: {OUTPUT_DIR}")


if __name__ == "__main__":
    main()
```

```python
import json
import os

def fix_null_lang(file_path, target_lang="akk-x-oldbab"):
    """
    Заменяет все записи с "lang": null на указанный язык.
    Сохраняет файл с тем же именем (перезапись).
    """
    print(f"Обработка: {file_path}")
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    # Если data — список словарей
    if isinstance(data, list):
        changed = 0
        for item in data:
            if item.get("lang") is None:
                item["lang"] = target_lang
                changed += 1
        print(f" Заменено {changed} записей на '{target_lang}'")
    else:
        print(" Ожидается список, но получен другой тип данных.")
        return

    # Сохраняеммм
    with open(file_path, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)
    print(f"  Сохранено в {file_path}\n")


if __name__ == "__main__":
    files = [
        "/home/kuklelik/Jupyter/data_html_akk/output2/obel.json",
        "/home/kuklelik/Jupyter/data_html_akk/output2/obmc.json"
    ]

    for fpath in files:
        if os.path.exists(fpath):
            fix_null_lang(fpath, target_lang="akk-x-oldbab")
        else:
            print(f"Файл не найден: {fpath}")
```

```python
import os
import json
import re
from collections import defaultdict
from pathlib import Path

# f очистки текста 
def clean_akkadian_text(text: str) -> list:
    """
    Удаляет мусор, но сохраняет аккадские буквы с диакритикой.
    """
    if not text or not isinstance(text, str):
        return []
    
    # Удаляем всё внутри квадратных скобок
    text = re.sub(r'\[.*?\]', ' ', text)
    
    # Удаляем угловые скобки
    text = re.sub(r'\<\<.*?\>\>', ' ', text)
    
    # Удаляем специальные клинописные знаки
    text = re.sub(r'[⸢⸣]', ' ', text)
    
    # Удаляем цифры (включая подстрочные/надстрочные типа qa₂-be-ki)
    text = re.sub(r'\d+', ' ', text)
    
    # Разбиваем на токены
    tokens = text.split()
    clean_tokens = []
    
    for token in tokens:
        # Удаляем x-x-x и подобное
        if re.search(r'x-|-x', token.lower()):
            continue
        # Удаляем недопустимые символы с краёв
        token = re.sub(r"^[^a-zA-Z\u0100-\u024F\u1E00-\u1EFF\u0250-\u02AF]+", "", token)
        token = re.sub(r"[^a-zA-Z\u0100-\u024F\u1E00-\u1EFF\u0250-\u02AF]+$", "", token)
        
        # Пропускаем короткие и мусорные токены
        if len(token) < 2:
            continue
        
        # Оставляем только буквы + диакритику + апостроф/дефис внутри
        cleaned = re.sub(r"[^a-zA-Z\u0100-\u024F\u1E00-\u1EFF\u0250-\u02AF'-]", "", token)
        cleaned = cleaned.strip("'-")
        
        if len(cleaned) >= 2 and cleaned not in {"u", "x"}:
            clean_tokens.append(cleaned)
    
    return clean_tokens


# Обработка файлов из output
def main():
    BASE_DIR = Path("/home/kuklelik/Jupyter")
    INPUT_DIR = BASE_DIR / "data_html_akk" / "output2"
    OUTPUT_DIR = BASE_DIR / "data_html_akk" / "output_final2"
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    # Собираем документы по языкам
    lang_documents = defaultdict(list)

    for filename in INPUT_DIR.glob("*.json"):
        print(f"Обработка {filename}...")
        try:
            with open(filename, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception as e:
            print(f" Ошибка чтения: {e}")
            continue

        for item in data:
            if not isinstance(item, dict):
                continue
            text = item.get("text", "")
            lang = item.get("lang", "")
            
            if not lang or not lang.startswith("akk-x-") or not text:
                continue
                
            lemmas = clean_akkadian_text(text)
            if lemmas:
                lang_documents[lang].append(lemmas)

    mapping = {
        "akk-x-mbperi": "mid_babylonian_peripheral_lemmatized.txt",
        "akk-x-midass": "mid_assyrian_lemmatized.txt",
        "akk-x-stdbab": "std_babylonian_lemmatized.txt",
        "akk-x-oldbab": "old_babylonian_lemmatized.txt",
        "akk-x-oldakk": "old_akkadian_lemmatized.txt",
        "akk-x-neoass": "neo_assyrian_lemmatized.txt",
        "akk-x-neobab": "neo_babylonian_lemmatized.txt",
        "akk-x-midbab": "mid_babylonian_lemmatized.txt"
    }

    total_docs = 0
    for lang_tag, filename in mapping.items():
        docs = lang_documents.get(lang_tag, [])
        if docs:
            filepath = OUTPUT_DIR / filename
            with open(filepath, "w", encoding="utf-8") as f:
                for lemmas in docs:
                    f.write(" ".join(lemmas) + "\n")
            print(f" Сохранено {len(docs)} документов в '{filename}'")
            total_docs += len(docs)
        else:
            print(f" Нет данных для {lang_tag}")

    print(f" Всего обработано {total_docs} документов.")
    print(f" Результаты в: {OUTPUT_DIR}")


if __name__ == "__main__":
    main()
```

```python
#atae

import os
import json
import sys
from collections import defaultdict


def extract_lemmas_and_lang_from_cdl(cdl_list, lemmas_by_lang):
    for item in cdl_list:
        if item.get("node") == "l":
            f_data = item.get("f", {})
            cf = f_data.get("cf")
            lang = f_data.get("lang", "unknown")

            if not cf or cf in ("[unclear]", "[missing]", "u", "x"):
                continue

            if lang and lang.startswith("akk-x-"):
                lemmas_by_lang[lang].append(cf)

        elif "cdl" in item:
            extract_lemmas_and_lang_from_cdl(item["cdl"], lemmas_by_lang)


def process_corpusjson_dir(corpusjson_dir):
    lang_documents = defaultdict(list)

    if not os.path.isdir(corpusjson_dir):
        return lang_documents

    for filename in os.listdir(corpusjson_dir):
        if not (filename.startswith("P") and filename.endswith(".json")):
            continue

        filepath = os.path.join(corpusjson_dir, filename)
        try:
            with open(filepath, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception as e:
            print(f"Ошибка при чтении {filepath}: {e}", file=sys.stderr)
            continue

        lemmas_by_lang = defaultdict(list)
        if "cdl" in data:
            extract_lemmas_and_lang_from_cdl(data["cdl"], lemmas_by_lang)

        for lang, lemmas in lemmas_by_lang.items():
            if lemmas:
                lang_documents[lang].append(lemmas)

    return lang_documents


def find_atae_corpusjson_paths(root_dir):
    paths = []
    for root, dirs, files in os.walk(root_dir):
        if os.path.basename(root) == "corpusjson":
            paths.append(root)
    return paths


if __name__ == "__main__":
    atae_root = "/home/kuklelik/Jupyter/atae/atae"
    atae_paths = find_atae_corpusjson_paths(atae_root)
    
    print(f"Найдено {len(atae_paths)} папок corpusjson в ATAE")
    if not atae_paths:
        print("Ничего не найдено — проверьте путь.")
        sys.exit(0)

    print("Обработка только ATAE...")
    lang_docs = defaultdict(list)
    for path in atae_paths:
        docs_by_lang = process_corpusjson_dir(path)
        for lang, docs in docs_by_lang.items():
            lang_docs[lang].extend(docs)

    print("\nНайденные языковые метки в ATAE:")
    for lang in sorted(lang_docs.keys()):
        print(f"  {lang}: {len(lang_docs[lang])} документов")

    # Сохраняем в файлы с _dop
    mapping = {
        "akk-x-neoass": "neo_assyrian_lemmatized_dop.txt",
        "akk-x-neobab": "neo_babylonian_lemmatized_dop.txt",
        "akk-x-stdbab": "std_babylonian_lemmatized_dop.txt",
        "akk-x-midass": "mid_assyrian_lemmatized_dop.txt"
    }

    for lang_tag, filename in mapping.items():
        docs = lang_docs.get(lang_tag, [])
        if docs:
            with open(filename, "w", encoding="utf-8") as f:
                for lemmas in docs:
                    f.write(" ".join(lemmas) + "\n")
            print(f"\n Сохранено {len(docs)} документов в '{filename}'")
        else:
            print(f"\n Нет данных для {lang_tag} → '{filename}' не создан")

    print("\nГотово! Только ATAE, без SAAo.")
```

```python
from datetime import datetime
import os
import sys
import time
import traceback
from collections import Counter
from pathlib import Path
from urllib.parse import urlparse
import json
from playwright.async_api import async_playwright  # - АСИНХРОННЫЙ
from bs4 import BeautifulSoup
import asyncio


def extract_lemmatized_text(html: str) -> dict:
    soup = BeautifulSoup(html, 'html.parser')
    table = soup.find('table', class_='transliteration')
    if not table:
        raise ValueError("Таблица транслитерации не найдена")

    parts = []
    langs = []

    for td in table.select('td.tlit'):
        td_copy = BeautifulSoup(str(td), 'html.parser').find('td')
        if not td_copy:
            continue

        for span in td_copy.find_all('span', class_='w'):
            title = span.get('title', '')
            lemma = title.split('[', 1)[0].strip() if '[' in title else span.get_text().strip()
            if lemma:
                span.replace_with(lemma)

            for cls in span.get('class', []):
                if cls.startswith(('akk-x-')):
                    langs.append(cls)
                    break

        cell_text = td_copy.get_text().strip()
        if cell_text:
            parts.append(cell_text)

    text = ' '.join(parts)
    text = ' '.join(text.split())
    lang = Counter(langs).most_common(1)[0][0] if langs else None
    return {"text": text, "lang": lang}


def save_exception_log(savepath: str, err: Exception, source: str, **kwargs):
    print(f'\n-------  ◀ save_exception_log ▶  source={source}  savepath={savepath})  -------\n')
    os.makedirs(os.path.join(savepath, 'errors'), exist_ok=True)
    now = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
    filename = f"{source}_{now}.json"
    full_savepath = os.path.join(savepath, 'errors', filename)

    error_type = type(err).__name__
    error_message = str(err)
    _, _, tb = sys.exc_info()
    tb_line = traceback.extract_tb(tb)[-1]
    log_data = {
        "source": source,
        "type": error_type,
        "message": error_message,
        'filename': tb_line.filename,
        'lineno': tb_line.lineno,
        'code': tb_line.line,
        'traceback': traceback.format_exc(),
        "timestamp": datetime.now().isoformat(),
    }
    if kwargs:
        for name, val in kwargs.items():
            log_data[name] = val

    os.makedirs(os.path.dirname(full_savepath), exist_ok=True)
    with open(full_savepath, "w", encoding="utf-8") as f:
        json.dump(log_data, f, ensure_ascii=False, indent=2)
    print(f"Ошибка сохранена в {full_savepath}")


async def run(source_url, savepath_output, savepath_data):
    source_url = source_url.strip()
    corpus_name = urlparse(source_url).path.strip('/').split('/')[-2] + '.json'

    try:
        async with async_playwright() as p:
            browser = await p.chromium.launch(headless=True)
            page = await browser.new_page()
            await page.goto(source_url)
            await page.wait_for_selector("table", timeout=10000)

            # Получаем все строки таблицы
            rows = await page.query_selector_all("table tr")
            corpus = []

            for i in range(len(rows)):
                cells = await rows[i].query_selector_all("td")
                if len(cells) >= 2:
                    link = await cells[1].query_selector("a")
                    if link:
                        await link.click()
                        await page.wait_for_load_state("networkidle")

                        try:
                            html = await page.content()
                            data = extract_lemmatized_text(html)
                            corpus.append(data)
                            print('кайф')
                        except Exception as err:
                            print(f'Почему то получилась хуйня: {page.url}')
                            print(err)

                        # Возвращаемся на список
                        await page.goto(source_url)
                        await page.wait_for_selector("table", timeout=10000)
                        # Обновляем список строк (DOM мог измениться)
                        rows = await page.query_selector_all("table tr")

            savepath_corpus = os.path.join(savepath_output, corpus_name)
            print(f'Корпус сохранён в {savepath_corpus}')
            with open(savepath_corpus, 'w', encoding='utf-8') as file:
                json.dump(corpus, file, ensure_ascii=False, indent=4)

            await browser.close()

    except Exception as err:
        save_exception_log(savepath_data, err, corpus_name, **{'Оляша': "Милашка"})
        print(f"Ошибка при подключении или работе: {err}")


async def main():
    BASE_DIR = Path.cwd()
    savepath_data = BASE_DIR / 'data_html_akk'
    savepath_output = savepath_data / 'output_download_per_html'
    os.makedirs(savepath_output, exist_ok=True)
    
    targets = [
#         "https://oracc.museum.upenn.edu/aemw/amarna/pager",
#         "https://oracc.museum.upenn.edu/akklove/corpus",
#         "https://oracc.museum.upenn.edu/glass/corpus",
        
#         "https://oracc.museum.upenn.edu/aemw/alalakh/idrimi/pager",
#         "https://oracc.museum.upenn.edu//asbp/pager",
#         "https://oracc.museum.upenn.edu/babcity/pager",
#         "https://oracc.museum.upenn.edu//balt/corpus",
#         "https://oracc.museum.upenn.edu//borsippa/corpus",
#         "https://oracc.museum.upenn.edu/cams/barutu/corpus",
#         "https://oracc.museum.upenn.edu/cmawro/pager",
#         "https://oracc.museum.upenn.edu/kish/corpus/",
#         "https://oracc.museum.upenn.edu/obel/pager",
#         "https://oracc.museum.upenn.edu/obmc/corpus",
        
#         "https://oracc.museum.upenn.edu/tcma/corpus",    
        
        "https://oracc.museum.upenn.edu/ario/corpus",
        "https://oracc.museum.upenn.edu/rinap/corpus",
        "https://oracc.museum.upenn.edu/ribo/corpus",
        "https://oracc.museum.upenn.edu/riao/corpus"
        
    ]

    for url in targets:
        await run(url, savepath_output, savepath_data)
```

```python
import os
import json
import re
from collections import defaultdict
from pathlib import Path

# f очистки текста 
def clean_akkadian_text(text: str) -> list:
    """
    Удаляет мусор, но сохраняет аккадские буквы с диакритикой.
    """
    if not text or not isinstance(text, str):
        return []
    
    # Удаляем всё внутри квадратных скобок
    text = re.sub(r'\[.*?\]', ' ', text)
    
    # Удаляем угловые скобки
    text = re.sub(r'\<\<.*?\>\>', ' ', text)
    
    # Удаляем специальные клинописные знаки
    text = re.sub(r'[⸢⸣]', ' ', text)
    
    # Удаляем цифры (включая подстрочные/надстрочные типа qa₂-be-ki)
    text = re.sub(r'\d+', ' ', text)
    
    # Разбиваем на токены
    tokens = text.split()
    clean_tokens = []
    
    for token in tokens:
        # Удаляем x-x-x и подобное
        if re.search(r'x-|-x', token.lower()):
            continue
        # Удаляем недопустимые символы с краёв
        token = re.sub(r"^[^a-zA-Z\u0100-\u024F\u1E00-\u1EFF\u0250-\u02AF]+", "", token)
        token = re.sub(r"[^a-zA-Z\u0100-\u024F\u1E00-\u1EFF\u0250-\u02AF]+$", "", token)
        
        # Пропускаем короткие и мусорные токены
        if len(token) < 2:
            continue
        
        # Оставляем только буквы + диакритику + апостроф/дефис внутри
        cleaned = re.sub(r"[^a-zA-Z\u0100-\u024F\u1E00-\u1EFF\u0250-\u02AF'-]", "", token)
        cleaned = cleaned.strip("'-")
        
        if len(cleaned) >= 2 and cleaned not in {"u", "x"}:
            clean_tokens.append(cleaned)
    
    return clean_tokens


# Обработка файлов из output
def main():
    BASE_DIR = Path("/home/kuklelik/Jupyter")
    INPUT_DIR = BASE_DIR / "data_html_akk" / "output_download_per_html"
    OUTPUT_DIR = BASE_DIR / "data_html_akk" / "output_download_per_html_final"
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    # Собираем документы по языкам
    lang_documents = defaultdict(list)

    for filename in INPUT_DIR.glob("*.json"):
        print(f"Обработка {filename}...")
        try:
            with open(filename, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception as e:
            print(f" Ошибка чтения: {e}")
            continue

        for item in data:
            if not isinstance(item, dict):
                continue
            text = item.get("text", "")
            lang = item.get("lang", "")
            
            if not lang or not lang.startswith("akk-x-") or not text:
                continue
                
            lemmas = clean_akkadian_text(text)
            if lemmas:
                lang_documents[lang].append(lemmas)

    mapping = {
        "akk-x-mbperi": "mid_babylonian_peripheral_lemmatized.txt",
        "akk-x-midass": "mid_assyrian_lemmatized.txt",
        "akk-x-stdbab": "std_babylonian_lemmatized.txt",
        "akk-x-oldbab": "old_babylonian_lemmatized.txt",
        "akk-x-oldakk": "old_akkadian_lemmatized.txt",
        "akk-x-neoass": "neo_assyrian_lemmatized.txt",
        "akk-x-neobab": "neo_babylonian_lemmatized.txt",
        "akk-x-midbab": "mid_babylonian_lemmatized.txt"
    }

    total_docs = 0
    for lang_tag, filename in mapping.items():
        docs = lang_documents.get(lang_tag, [])
        if docs:
            filepath = OUTPUT_DIR / filename
            with open(filepath, "w", encoding="utf-8") as f:
                for lemmas in docs:
                    f.write(" ".join(lemmas) + "\n")
            print(f" Сохранено {len(docs)} документов в '{filename}'")
            total_docs += len(docs)
        else:
            print(f" Нет данных для {lang_tag}")

    print(f" Всего обработано {total_docs} документов.")
    print(f" Результаты в: {OUTPUT_DIR}")


if __name__ == "__main__":
    main()
```

```python
import os
import glob
import shutil

# Пути
base_dir = "/home/kuklelik/Jupyter"
final_output_dir = os.path.join(base_dir, "final_akk")

# Список целевых языковых файлов
lang_files = [
    "neo_assyrian_lemmatized",
    "neo_babylonian_lemmatized",
    "std_babylonian_lemmatized",
    "mid_assyrian_lemmatized",
    "old_babylonian_lemmatized",
    "mid_babylonian_peripheral_lemmatized",
    "old_akkadian_lemmatized",
    "mid_babylonian_lemmatized"
]

# Источники
sources = [
    base_dir,  # содержит *_lemmatized.txt и *_dop.txt
    os.path.join(base_dir, "data_html_akk", "output_final"),
    os.path.join(base_dir, "data_html_akk", "output_final2"),
    os.path.join(base_dir, "data_html_akk", "output_download_per_html_final")
]

# Создаём выходную папку
os.makedirs(final_output_dir, exist_ok=True)

# Для каждого языка — объединяем все подходящие файлы
for lang_base in lang_files:
    all_lines = []
    
    # Ищем все файлы: lang_base + (".txt" или "_dop.txt")
    for source in sources:
        if not os.path.isdir(source):
            continue
        
        # Основной файл
        main_path = os.path.join(source, f"{lang_base}.txt")
        if os.path.isfile(main_path):
            with open(main_path, "r", encoding="utf-8") as f:
                all_lines.extend(f.readlines())
        
        # Дополнительный файл (_dop.txt)
        dop_path = os.path.join(source, f"{lang_base}_dop.txt")
        if os.path.isfile(dop_path):
            with open(dop_path, "r", encoding="utf-8") as f:
                all_lines.extend(f.readlines())
    
    # Сохраняем объединённый файл
    if all_lines:
        output_path = os.path.join(final_output_dir, f"{lang_base}.txt")
        with open(output_path, "w", encoding="utf-8") as f:
            f.writelines(all_lines)
        print(f"Объединено {len(all_lines)} строк → {output_path}")
    else:
        print(f"Нет данных для {lang_base}")

print("Все корпуса объединены в:", final_output_dir)
```

```python
import os
import re

# Путь к папке с финальными корпусами
input_dir = "/home/kuklelik/Jupyter/final_akk"

# Получаем все .txt файлы
for filename in os.listdir(input_dir):
    if not filename.endswith(".txt") or "_filtered" in filename:
        continue

    input_path = os.path.join(input_dir, filename)
    output_path = os.path.join(input_dir, filename.replace(".txt", "_filtered.txt"))

    filtered_lines = []

    with open(input_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue

            # Убираем дефисы и заменяем их на пустоту (слова склеиваются)
            line = line.replace("-", "")

            # Переводим в нижний регистр
            line = line.lower()

            # Разбиваем на слова (по пробелам)
            words = line.split()

            # Оставляем только если >= 3 слов
            if len(words) >= 3:
                filtered_lines.append(" ".join(words) + "\n")

    # Сохраняем результат
    with open(output_path, "w", encoding="utf-8") as f:
        f.writelines(filtered_lines)

    print(f"Отфильтровано: {filename} → {len(filtered_lines)} строк")

print("Фильтрация завершена!")
```
